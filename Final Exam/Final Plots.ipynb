```python
import matplotlib.pyplot as plt
import pandas as pd

# Data extracted from Minard's map_
data = {
    "City": [
        "Kovno", "Vilna", "Glubokoye", "Smolensk", "Dorogobuzh", "Chjat", "Moscow", 
        "Moscow", "Vyazma", "Orsha", "Bobr", "Minsk", "Smorgon", "Kovno"
    ],
    "Longitude": [24, 25.3, 27, 32, 33.5, 36, 37.6, 37.6, 34.5, 30, 29, 27, 26, 24],
    "Soldiers": [422000, 400000, 60000, 145000, 55000, 127100, 100000, 100000, 37000, 20000, 20000, 28000, 12000, 10000],
    "Direction": [
        "Advance", "Advance", "Advance", "Advance", "Advance", "Advance", "Advance",
        "Retreat", "Retreat", "Retreat", "Retreat", "Retreat", "Retreat", "Retreat"
    ]
}

df_minard = pd.DataFrame(data)

plt.figure(figsize=(14, 6))
for direction, color in zip(["Advance", "Retreat"], ["#7FB3D5", "#F5B041"]):
    segment = df_minard[df_minard["Direction"] == direction]
    plt.plot(segment["Longitude"], segment["Soldiers"], marker='o', linewidth=4, label=direction, color=color)

plt.title("Napoleon’s Russian Campaign")
plt.xlabel("Longitude (Approximate)")
plt.ylabel("Number of Soldiers")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
D
# Generate fake data
np.random.seed(42)
group1 = np.concatenate([np.random.normal(50, 2, 50), np.random.normal(70, 2, 50)])
group2 = np.concatenate([np.random.normal(55, 2, 50), np.random.normal(75, 2, 50)])

df = pd.DataFrame({
    'Outcome': np.concatenate([group1, group2]),
    'Group': ['Group 1'] * len(group1) + ['Group 2'] * len(group2)
})

plt.figure(figsize=(8, 6))
sns.boxplot(x='Group', y='Outcome', data=df)
sns.stripplot(x='Group', y='Outcome', data=df, color='black', size=3, jitter=True, alpha=0.4)
plt.title('Box Plot Fails to Show Bimodal Distributions')
plt.tight_layout()
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

dances = [
    "Bharatanatyam", "Kathak", "Odissi", "Kathakali", "Kuchipudi", "Manipuri", "Mohiniyattam", "Sattriya", "Yakshagana", "Chhau"
]

edges = [
    ("Bharatanatyam", "Kuchipudi"),
    ("Kathak", "Odissi"),
    ("Kathakali", "Mohiniyattam"),
    ("Sattriya", "Manipuri"),
    ("Yakshagana", "Kathakali"),
    ("Chhau", "Odissi"),
    ("Kuchipudi", "Mohiniyattam"),
    ("Bharatanatyam", "Mohiniyattam"),
    ("Kathak", "Sattriya"),
    ("Yakshagana", "Chhau"),
    ("Bharatanatyam", "Kathakali"),
    ("Kathak", "Manipuri"),
    ("Kuchipudi", "Odissi"),
    ("Sattriya", "Mohiniyattam"),
    ("Manipuri", "Odissi")
]

G = nx.Graph()
G.add_nodes_from(dances)
G.add_edges_from(edges)

plt.figure(figsize=(10, 7))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx_nodes(G, pos, node_size=1000, node_color='lightblue')
nx.draw_networkx_edges(G, pos, width=2, alpha=0.6)
nx.draw_networkx_labels(G, pos, font_size=10, font_weight='bold')
plt.title("Network of Classical Indian Dance Forms", fontsize=14)
plt.axis("off")
plt.tight_layout()
plt.show()
```
