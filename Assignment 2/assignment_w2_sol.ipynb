```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")

data = pd.read_json('sol_data.json')
data.head()
```

### 1. General Information

- How many objects are in the dataset?
- How many are planets? How many are moons?


```python
# Total number of objects
total_objects = len(data)
print(f"Total Number of Objects: {total_objects}")

# Number of planets
num_planets = data[data['isPlanet'] == True].shape[0]
print(f"Total Number of Planets: {num_planets}")

# Number of moons (Secondary objects)
num_moons = data[data['orbit_type'] == 'Secondary'].shape[0]
print(f"Total Number of Moons: {num_moons}")
```

> **Hint**: By moon we mean a natural satellite of a planet or another object in the solar system. Take a look at the columns and see if you can identify the criteria for classifying an object as a moon. Ask the instructor or AIs for help if needed. 

### 2. Planets

- What is the mean density of all planets?
- Which planet has the highest surface gravity, and what is its gravity value?
- List all planets in descending order of their mass.


```python
planets = data[data['isPlanet'] == True]

# Mean density of all planets
mean_density = planets['density'].mean()
print(f"Mean density of all planets: {mean_density:.2f} kg/m³")

# Planet with the highest surface gravity
planet_highest_gravity = planets.loc[planets['gravity'].idxmax(), ['eName', 'gravity']]
print(f"Planet with the highest surface gravity: {planet_highest_gravity['eName']} ({planet_highest_gravity['gravity']} m/s²)")

# Planets sorted by descending mass
planets_sorted_by_mass = planets.sort_values(by='mass_kg', ascending=False)
print("\nPlanets sorted by descending mass:")
print(planets_sorted_by_mass[['eName', 'mass_kg']])
```

### 3. Moons
- How many moons orbit each planet? Present this as a table or dictionary.
- What is the average radius (meanRadius) of all moons?
- Compare the average surface gravity of moons to that of planets.


```python

moons = data[data['orbit_type'] == 'Secondary']

# Number of moons orbiting each planet
moons_per_planet = moons.groupby('orbits')['eName'].count().reset_index()
moons_per_planet.columns = ['Planet', 'Number of Moons']

all_planets = planets[['eName']].copy()
all_planets = all_planets.rename(columns={'eName': 'Planet'})
moons_per_planet = all_planets.merge(moons_per_planet, on='Planet', how='left').fillna(0)

moons_per_planet['Number of Moons'] = moons_per_planet['Number of Moons'].astype(int)
print("Number of moons orbiting each planet:")
print(moons_per_planet)

# Average radius of all moons
average_moon_radius = moons['meanRadius'].mean()
print(f"\nAverage radius of all moons: {average_moon_radius:.2f} km")

# Compare average surface gravity of moons vs. planets
avg_gravity_planets = planets['gravity'].mean()
avg_gravity_moons = moons['gravity'].mean()

print(f"\nAverage surface gravity of moons: {avg_gravity_moons:.2f} m/s²")
print(f"Average surface gravity of planets: {avg_gravity_planets:.2f} m/s²")
```

### 4. Orbital Properties

- Which object has the highest orbital eccentricity, and what is its value?
- Calculate the average semi-major axis (semimajorAxis) for planets and compare it to that of moons.
- Identify the moon with the shortest orbital period (sideralOrbit) and the planet it orbits.


```python
# Highest orbital eccentricity
highest_eccentricity = data.loc[data['eccentricity'].idxmax(), ['eName', 'eccentricity']]
print(f"Object with the highest orbital eccentricity: {highest_eccentricity['eName']} ({highest_eccentricity['eccentricity']:.4f})")

# Average semi-major axis of planets vs. moons
avg_semimajor_planets = planets['semimajorAxis'].mean()
avg_semimajor_moons = moons['semimajorAxis'].mean()

print(f"\nAverage semi-major axis of planets: {avg_semimajor_planets:.2f} km")
print(f"Average semi-major axis of moons: {avg_semimajor_moons:.2f} km")

# Moon with the shortest orbital period
moon_shortest_orbit = moons.loc[moons['sideralOrbit'].abs().idxmin(), ['eName', 'sideralOrbit']]
print(f"\nMoon with the shortest orbital period: {moon_shortest_orbit['eName']} ({moon_shortest_orbit['sideralOrbit']} days)")
```

### 5. Discovery Dates

- How many objects have recorded discovery dates?
- Which is the oldest discovered moon, and when was it discovered?


```python
# Objects with discovery dates
discovered_objects = data[data['discoveryDate'].notna()]
print("Objects with recorded discovery dates:", len(discovered_objects))

# Oldest discovered moon
moons = moons.copy()
moons.loc[:, 'discoveryDate'] = pd.to_datetime(moons['discoveryDate'], errors='coerce')

moons_notna = moons.dropna(subset=['discoveryDate'])

oldest_moon = moons_notna.loc[moons_notna['discoveryDate'].idxmin(), ['eName', 'discoveryDate']]
print(f"Oldest discovered moon: {oldest_moon['eName']} (Discovered on {oldest_moon['discoveryDate'].date()})")

# I used pd.to_datetime for the conversion, but it discards any dates before the year 1700. As a result, the moon date from 1787 is shown as the oldest in the output.

```

### 6. Advanced Analysis

- Calculate the average density of moons that orbit planets with a mass greater than Earth's mass (`5.97e24 kg`).
- Group all objects by their `orbit_type` and compute the average orbital eccentricity for each group.
- Identify the top 3 moons with the highest escape velocity (escape).


```python
# Average density of moons orbiting planets with mass > Earth

mass_threshold = 5.972e24
massive_planets = data[(data['isPlanet'] == True) & (data['mass_kg'] > mass_threshold)]

moons_orbiting_massive_planets = data[(data['orbit_type'] == 'Secondary') & (data['orbits'].isin(massive_planets['eName']))]
avg_density_moons = moons_orbiting_massive_planets['density'].mean()
print(f"Average density of moons orbiting planets with mass > Earth: {avg_density_moons:.2f}")

# Average orbital eccentricity by orbit_type
avg_eccentricity = data.groupby('orbit_type')['eccentricity'].mean()
print("\nAverage orbital eccentricity by orbit type:")
print(avg_eccentricity)

# Top 3 moons with highest escape velocity
top_3_moons_escape_velocity = data[data['orbit_type'] == 'Secondary'].nlargest(3, 'escape')[['eName', 'escape']]
print("\nTop 3 moons with highest escape velocity:")
print(top_3_moons_escape_velocity)
```

### 7. Extra questions

1. How many moons have a mass less than 10% of Earth's moon? What percentage of all moons does this represent?
2. Calculate the ratio of moons to planets in the dataset. Which planet has the highest number of moons relative to its mass?
3. Group moons by their host planet and calculate the average density for each group. Which planet hosts moons with the highest average density?

```python
earth_moon_mass = 7.35e22 * 0.1

# Moons with a mass less than Earth's moon and percentage
moons_less_than_earth_moon = moons[moons['mass_kg'] < earth_moon_mass]

percentage_moons = (len(moons_less_than_earth_moon) / len(moons)) * 100 if len(moons) > 0 else 0

print(f"Number of moons with mass less than Earth's Moon: {len(moons_less_than_earth_moon)}")
print(f"Percentage of moons with mass less than Earth's Moon: {percentage_moons:.2f}%")

# Ratio of moons to planets and planet with highest moon to mass ratio
moon_to_planet_ratio = len(moons) / len(planets) if len(planets) > 0 else 0

moons_per_planet = moons.groupby('orbits').size().reset_index(name='moon_count')
planets_with_moons = planets.merge(moons_per_planet, left_on='eName', right_on='orbits', how='left').fillna(0)
planets_with_moons['moon_to_mass_ratio'] = planets_with_moons['moon_count'] / planets_with_moons['mass_kg']

highest_moon_mass_ratio = planets_with_moons.loc[planets_with_moons['moon_to_mass_ratio'].idxmax(), ['eName', 'moon_to_mass_ratio']]

print(f"\nRatio of moons to planets: {moon_to_planet_ratio:.2f}")
print(f"Planet with highest moon-to-mass ratio: {highest_moon_mass_ratio['eName']}")

# Average density of moons per planet

avg_density_per_planet = moons.groupby('orbits')['density'].mean().reset_index()
avg_density_per_planet.columns = ['Planet', 'Average Moon Density']

all_planets = planets[['eName']].copy()
all_planets = all_planets.rename(columns={'eName': 'Planet'})
avg_density_per_planet = all_planets.merge(avg_density_per_planet, on='Planet', how='left').fillna(0)

avg_density_per_planet['Average Moon Density'] = avg_density_per_planet['Average Moon Density'].astype(int)

print("\nAverage density of moons per planet:")
print(avg_density_per_planet)

highest_density_planet = avg_density_per_planet.loc[avg_density_per_planet['Average Moon Density'].idxmax()]
print(f"\nPlanet with the highest average moon density: {highest_density_planet['Planet']} ({highest_density_planet['Average Moon Density']})")

```
