```python
import igraph as ig
import matplotlib.pyplot as plt

G_main = ig.Graph.Read_GML("Wiki_Science_small.gml")
G_ig = G_main.subgraph(range(1000))

layout = G_ig.layout("fr")  
fig, ax = plt.subplots(figsize=(12, 8))
ig.plot(
    G_ig,
    target=ax,
    layout=layout,
    vertex_size=8,
    vertex_label=None,
    edge_color="gray"
)
plt.title("Wiki Science Network (iGraph)")
plt.show()
```
