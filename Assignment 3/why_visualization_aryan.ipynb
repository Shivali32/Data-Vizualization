## Anscombe's Quartet dataset

````python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset
data = pd.read_csv('Anscombe_quartet_data.csv')
# The ../../ are needed to go back two levels in the directory structure.
# Note that the path is relative to the location of the notebook file.
````

Explore the dataset and calculate the mean and standard deviation for each dataset.

````python
# Explore here
x_col = ["x123", "x4"]
y_col = ["y1", "y2", "y3", "y4"]

x_mean = data[x_col].mean().mean()
y_mean = data[y_col].mean().mean()

print(f"Mean of X for x123 and x4: {x_mean}")
print(f"Mean of Y for y1, y2, y3, y4:{y_mean: .2f}")

x_all = pd.concat([data["x123"], data["x4"]], ignore_index=True)
x_std = x_all.std()
y_all = pd.concat([data["y1"], data["y2"], data["y3"], data["y4"]], ignore_index=True)
y_std = y_all.std()

print(f"Standard Deviation of x123 and x4:{x_std: .2f}")
print(f"Standard Deviation of y1, y2, y3, y4:{y_std: .2f}")
````

Also calculate the correlation between the x123 and y1,y2,y3 variables. Also do the same for the x4 and y4 variables.


````python
# Calculate correlations here
corr_xny_123 = data[["x123", "y1", "y2", "y3"]].corr()
corr_xny_4 = data[["x4", "y4"]].corr()

print("x123 and y1, y2, y3 Correlation:\n", corr_xny_123)
print("\nx4 and y4 Correlation:\n", corr_xny_4)
````

````python
correlation_matrix = data.corr()

plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, center=0, fmt=".2f")
plt.title("Correlation Heatmap for Entire Dataset")
plt.show()
````

````python
# Plot the data here

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
fig.suptitle("Datasets Visualizations")

axes[0, 0].scatter(data["x123"], data["y1"])
axes[0, 0].set_title("Dataset I: x1 vs y1")
axes[0, 0].set_xlabel("x1")
axes[0, 0].set_ylabel("y1")

axes[0, 1].scatter(data["x123"], data["y2"])
axes[0, 1].set_title("Dataset II: x2 vs y2")
axes[0, 1].set_xlabel("x2")
axes[0, 1].set_ylabel("y2")

axes[1, 0].scatter(data["x123"], data["y3"])
axes[1, 0].set_title("Dataset III: x3 vs y3")
axes[1, 0].set_xlabel("x3")
axes[1, 0].set_ylabel("y3")

axes[1, 1].scatter(data["x4"], data["y4"])
axes[1, 1].set_title("Dataset IV: x4 vs y4")
axes[1, 1].set_xlabel("x4")
axes[1, 1].set_ylabel("y4")

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
````

````python
import scipy.stats as stats

data = pd.read_csv('data9b.csv')
data.head()
````

````python
# Your code here

gender_female = data[data['gender'] == 'female']
gender_male = data[data['gender'] == 'male']

t_stat, p_mean_steps = stats.ttest_ind(gender_female['steps'], gender_male['steps'])
corr_female_bmi_steps, p_female_corr_value = stats.pearsonr(gender_female['steps'], gender_female['bmi'])
corr_male_bmi_steps, p_male_corr_value = stats.pearsonr(gender_male['steps'], gender_male['bmi'])

print("P-value for hypothesis")
print(f"(a) Difference in mean steps: {p_mean_steps}")
print(f"(b) Negative correlation for women: {p_female_corr_value}")
print(f"(c) Positive correlation for men: {p_male_corr_value}")
````
